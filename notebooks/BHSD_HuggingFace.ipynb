```python
path = dataset['test']._data.column('nifti')[0][1].as_py()

# split
zip_path = path.split("::")[1]
inner_path = path.split("::")[0].replace("zip://", "")
```

```python
import zipfile

with zipfile.ZipFile(zip_path, 'r') as z:
    with z.open(inner_path) as f:
        with open("sample.nii.gz", "wb") as out:
            out.write(f.read())
```
